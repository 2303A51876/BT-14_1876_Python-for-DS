```python
# LAB-8 ASSIGNMENT-1
# Student Performance Visualization

import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
data = pd.read_csv("student_marks.csv")

# Plot bar chart
plt.figure(figsize=(8,6))
plt.bar(data["Subject"], data["Average_Marks"], color=['skyblue','lightgreen','salmon','orange','violet','gold'])
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.title("Average Marks by Subject")
plt.show()

# Insights
max_subject = data.loc[data['Average_Marks'].idxmax(), 'Subject']
min_subject = data.loc[data['Average_Marks'].idxmin(), 'Subject']
print(f"Easiest subject (highest avg): {max_subject}")
print(f"Toughest subject (lowest avg): {min_subject}")
```

```python
# LAB-8 ASSIGNMENT-2
# Retail Sales Visualization

import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
data = pd.read_csv("sales_category.csv")

# Plot bar chart
plt.figure(figsize=(8,6))
plt.bar(data["Category"], data["Sales"],
        color=['red','blue','green','purple','orange','cyan'])
plt.xlabel("Product Category")
plt.ylabel("Sales")
plt.title("Sales by Product Category")
plt.show()

# Insights
best_category = data.loc[data['Sales'].idxmax(), 'Category']
print(f"Best-selling category: {best_category}")
```
